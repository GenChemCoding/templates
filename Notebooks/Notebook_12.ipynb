# Calculating pH

#ph = -log[H^+] # [H^+] is the concentration of the hydrogen ions
#molarity = n mol/vol
#[H+][OH-] = 1*10^-14

The formula for the pH calculation of weak acids and bases is derived from the equation:

# K = [H^+][A]/[HA] 

where A is some element that makes an acid ,

The ICE Chart for the reaction is as follows:

|  | HA | H^+ | A |
| --- | --- | --- | --- |
| Initial | C | 0| 0|
| Change| -x | x | x |
| Equilibirum| C - x| x | x|

Plugging in the results from chart yield:

## K = [x][x]/[C - x] --> x^2 = K(C-x) --> x^2 + KX - KC = 0

```python
def get_pH_weak_base(K,C):
    # Calculate the discriminant
    d = (K**2)-(4*K*C*-1)
    
    # Apply the quadratic formula
    root = ((-K + math.sqrt(d)) / (2))
    return 14 - -math.log(root,10) 

get_pH_weak_base(5.4*10**-4, 1)
```

```python
def get_pH_strong_acid(C):
    pH = -math.log(C,10)
    return pH
get_pH_strong_acid(0.1)
```
