### Creating a SMILE string:

SMILE strings are a way to condense molecules into a single string.

1. When representing a molecule as a SMILE string ignore Hydrogen atoms.
2. Two atoms next to each other, like CC, suggest a single bond.
3. A "=" denotes a double bond, a "#" denotes a triple bond.
4. A parenthesis between atoms like C(O)C suggests a branch of off the atom to the left.

````python
smiles = 'N'
mol = Chem.MolFromSmiles(smiles)
Draw.MolToImage(mol)
````
