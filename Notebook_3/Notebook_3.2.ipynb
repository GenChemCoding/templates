## Data Structures 

Data structres are the ways you can store information

The most common types are:

- String : str(true)

Whenever you type letter in python its assumed you either assigning a variable or calling a function eg. print(), str makes sure that it is interpreted as a set of characters, you can also do this simply adding quotations around a word, "True"

Example use: print(str(True)) 

Although this is an example where you can simply use ' ' to denote characrters str() can be used in other scenarios where the list of characters you want to print is variable.

- List : [value1, value2]

A list stores comma seperated values, they can be strings or integers. 

Example use: my_list = [], my_list.append('True'), my_list.append('False')

print(my_list) will show ['True','False'] 

You can also use the index to print a single specific value, for example, print(my_list[0]) will show only ['True'] because python counts starting from 0.

- Dictionary : { Key1 : Value1,  Key2 : Value2,  }

A dictionary is a made of what are called key-value pairs, they can either be intergers or strings

Example use: my_dict = {} , my_dict['key1'] = ['value1']

print(my_dict) will show {'key1': 'value1'}

- Tuple : (value1, value2)

A tuple holds 2 immutable values. Immutable means it cannot be changed after it is created.

Example use: my_tuple = (1,2)

Attempting to change a value by doing my_tuple[0] = 2 will cause an error.

```python
from tabulate import tabulate

# Define your data as a list of lists or dictionaries
data = [
    ["Alice", 24, "New York"],
    ["Bob", 19, "Los Angeles"],
    ["Charlie", 32, "Chicago"]
]
headers = ["Name", "Age", "City"]

# Print the table using the "grid" format
print(tabulate(data, headers=headers, tablefmt="grid"))
```
