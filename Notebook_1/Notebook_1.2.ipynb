We can also create lines of code to carry out specific processes which we can then call upon later in our programming. When we want to do this, we set the sequence of code to a <u>function</u>. So let's try to use these operations to create a function that will convert temperatures.

  The formula for converting celsius to farenheit is <font color="A7E8D1">(X°C × 9/5) + 32 = Y°F</font>. For farenheit to celsius, the equation is <font color="A7E8D1">(Y°F − 32) × 5/9 = X°C</font>. Let's try to use this formula.

````python
Temperature = 54
Measurement = 'C'
def convert_temp(Temperature, Measurement):
  if Measurement == 'C':
    Temperature = (Temperature * 9/5) + 32
    Measurement = 'F'
  elif Measurement == 'F':
    Temperature = (Temperature - 32) * 5/9
    Measurement = 'C'
  else:
    print('Invalid Measurement')
  return round(Temperature), Measurement
print(f"{Temperature:.1f} {Measurement}")
convert_temp(Temperature, Measurement)
````

### Rounding

In Python, there are built in ways in which we can round a number. Take a look at the code cell below:

````python
result = 13.365
print(f"{result:.2f}")
#.2f rounds the number to 2 decimal places which results in 4 significant figures.
````

In the code cell above, the line ``` print(f"{result:.2f}")``` is used to round the number 13.365 set to the variable ```result``` to the nearest 2 decimal places. This is useful when calculating significant figures in an equation as your starting number may have a different number of significant figures as opposed to the second, so adjusting the answer to match the significant figures of the first number can be made easier with this feature.

There are other ways we can round numbers in Python for precision. For rounding, you may also use the ```round()``` function in order to round a number to the nearest integer you indicate. For instance:

````python
result = 13.365
print(round(result, 2))
````

The ```round()``` function works in a slightly different way from the ```print(f{...:.2f})``` function as the first drops trailing zeros and saves the newly rounded value in the variable. The second function differs in that it outputs a string, but does not change the value of the variable to the output that has been rounded. Both can be useful depending on the context you're working in. When solving a problem, it is up to your discretion which will prove most useful.

````python
result = 27000.0
print(f"{result:.1e}")
# e+ is the same as *10^(n) and e- would be *10^(-n).
````
