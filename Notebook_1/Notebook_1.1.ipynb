## Significant Figures

Let's look at significant figures. Significant figures refer to the digits in a measurement. There are some rules we must abide to when looking at significant figures.

|Type of digit | Rule|
|:--:          | :--: |
|Non-zero integers| Always significant|
|Captive zero  | Always significant|
|Trailing with decimal point| Always significant |

|Leading zeros|Never significant|

## Dimensional Analysis

Dimensional analysis is a way in mathematics for people to keep track of varying units in a problem. Here is an example of this:

$\frac{25m}{10m/s} = 2.5 s$

Here we have two different units (meters and meters/seconds) being used in this operation. Let's take a closer look at this:

$\frac{meters}{meters/seconds} = seconds$

The process of going from meters divided by meters per second can be expanded like this:

$\frac{meters}{meters/seconds} = \frac{meters}{meters * 1/seconds} = \frac{1}{1/seconds} = 1 * \frac{seconds}{1}$ = seconds
