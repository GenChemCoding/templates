## 2.1.1 - Evolution of the Atomic Model

To understand chemistry and all of its concepts, it is first important to learn about the history of the atomic model and how we came about to understanding the way an atom is structured.

Dalton's Atomic Theory:
 <div style="margin-left: 2em;">Dalton's atomic theory described the concepts of matter, atoms, and how they interact in 5 postulates:
<br>

1. Matter is composed of exceedingly small particles called atoms; An atom is the smallest unit of an element that can participate in a chemical change
1. An element consists of only one type of atom, which has a mass that is characteristic of the element and is the same for all atoms of the same element 
1. Atoms of one element differ in properties from atoms of another element
1. A compound consists of atoms of two or more elements combined
1. Atoms are neither created nor destroyed; Instead, they are rearranged

Laws of definite and Multiple Proportions:
    Joseph Proust defined the law of definite proportions, or the law of constant composition, in which all samples of a pure comound contain the same elements in the same proportion by mass. This, however, does not mean that compounds with the same mass ratios are of the same substance. This phenomenon refers to the law of multiple proportions, where if two elements react to form more than one compound, a fized mass of one element will react with the masses of the other element in a ration of small whole numbers.

The Cathode Ray:
<div style="margin-left: 2em;">J. J. Thomson used the cathode ray tube to investigate the electrical charges produced in low pressure gases. Thomson had proposed that the particles were attracted to each other by positive charges and repelled by negative charges, concluding that these particles had a negative charge. These particles were determined to be essential subatomic particles of the atom, which is now called the electron.</div>

## 2.1.2 - Atomic Structure and Chemical Formulas

The basic atomic structure describes the atom to have protons and neutrons in the nucleus, with a cloud of electrons surrounding the outskirts of this nucleus. 

Atoms are very small, having a diameter of 10 <sup>-10</sup> m. In comparison, the nucleus has a diameter of 10 <sup>-15</sup> m. The size of the atom and its subatomic structures are very, very tiny, which led to the creation of the unified atomic mass unit (u) and the fundamental unit of charge (e) to be used to describe their physical properties. 

The fundamental unit charge is equal to the magnitude of an electron (e = 1.602 x 10<sup>-19</sup>). The unified atomic mass unit is exactly 1/12 of the mass of a single atom of the carbon-12 isotope, which is equal to 1.6605 x 10 <sup> -24 </sup> g. a Dalton (Da) is an alternative mass unit that is equivalent to the unified atomic mass unit. The older term for the dalton and unified atomic mass unit is the atomic mass unit (amu).
