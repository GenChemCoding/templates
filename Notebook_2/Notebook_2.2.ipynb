## 2.2.1 - What is a Function?

### Functions:

In mathematics, functions have a single input, which give one output. Functions can provide useful information on a problem that can then be applied to new functions, providing even more information. Some equations you may be familiar with that are functions include the Pythagorean theorem, sin(x), and f(x) = e<sup>x</sup>. 

In python, functions are used for the same reasons: getting outputs that will help us execute certain actions in another functions. The main difference between the two is the amount of inputs and outputs you get from a mathematical function v.s. a function in python. With mathematical functions, in order to be considered a proper function, they must have exactly one input and one output. In python, you can have zero, one, or multiple inputs and one or more outputs.

We can write code to carry out certain actions we need the computer to do for us. But often times, these blocks of code can be very long. And if we need to carry out these same computations later in our program, it's a lot to copy and paste the entire block of code we need to carry out the action or actions we need. So to solve this, we can use a function to save the lines of code and then call them later. In this section, we'll better describe what functions are, how to use them, and how to make them.

A module is a file that contains one or more functions, which can then be imported into your workspace.
