<font color="8DAEF0">&ensp;Let's start with mass v.s. weight. <font color="F09D8D">Mass</font> refers to the amount of matter within an object, while <font color="F09D8D">weight</font> refers to the force at which an object is being pulled to the Earth's core by gravity. So what's matter? <font color="F09D8D">Matter</font> is anything that occupies a space or has a mass.

&ensp;Next, let's review the states of matter. There are 4 states of matter: solid, liquid, gas, and plasma. For our purposes, we will only focus on the first 3. <font color="F09D8D">Solid</font> matter is rigid  with a definite shape, <font color="F09D8D">liquids</font> flow and take the shape of their container, and <font color="F09D8D">gases</font> assume the shape of their container while filling its volume.</font>
