<font color="8DAEF0"> The formula for converting celsius to farenheit is <font color="A7E8D1">(X°C × 9/5) + 32 = Y°F</font>. For farenheit to celsius, the equation is <font color="A7E8D1">(Y°F − 32) × 5/9 = X°C</font>. Let's try to use this formula.

````python
Temperature = 54
Measurement = 'C'
def convert_temp(Temperature, Measurement):
  if Measurement == 'C':
    Temperature = (Temperature * 9/5) + 32
    Measurement = 'F'
  elif Measurement == 'F':
    Temperature = (Temperature - 32) * 5/9
    Measurement = 'C'
  else:
    print('Invalid Measurement')
  return round(Temperature), Measurement
print(f"{Temperature:.1f} {Measurement}")
convert_temp(Temperature, Measurement)
````

In Python, there are built in ways in which we can round a number. Take a look at the code cell below:

This is a useful way to truncate values to be approximations that we need for specific contexts.

````python
result = 27000.0
print(f"{result:.1e}")
# e+ is the same as *10^(n) and e- would be *10^(-n).
````

Here we have 17.6745 rounded to 3 significant figures:

````python
result = 17.6745
print(f"{result:.1f}")
````

````python
import math
num = 0.0053030
sig_figs = 3
def round_to_sig_figs(num, sig_figs):
    """
    Rounds a number to a specified number of significant figures.

    Args:
        num (float or int): The number to round.
        sig_figs (int): The desired number of significant figures.

    Returns:
        float: The number rounded to the specified significant figures.
    """
    if num == 0:
        return 0.0

    # Calculate the exponent of the number in scientific notation (e.g., for 17.6, exponent is 1)
    # For 0.0012, exponent is -3.
    exponent = math.floor(math.log10(abs(num)))

    # Calculate the precision needed for Python's built-in round() function.
    # round() rounds to a number of decimal places. We need to adjust based on the exponent.
    # For 17.6745 and 3 sig figs:
    #   exponent = 1
    #   precision = 3 - 1 - 1 = 1 (round to 1 decimal place)
    # For 0.001234 and 2 sig figs:
    #   exponent = -3
    #   precision = 2 - (-3) - 1 = 4 (round to 4 decimal places)
    precision = sig_figs - (int(exponent) + 1) # (int(exponent) + 1) is calculating how many
    # numbers are above the decimal. E.G. 17.6 would be 1.76 * 10^1  1+1 = 2; 176.5 would be 1.76*10^2 2+1 = 3.
    # Then the number of digits above the decimal point are subtracted from the total sig figs to find what to round to
    return round(num, precision)

# Example usage:
#print(round_to_sig_figs(17.67, 3)) # Expected: 17.7
# print(round_to_sig_figs(0.001234, 2)) # Expected: 0.0012
# print(round_to_sig_figs(12345, 2)) # Expected: 12000.0
# print(round_to_sig_figs(12.0, 3)) # Expected: 12.0
# print(round_to_sig_figs(99.99, 3)) # Expected: 100.0
# print(round_to_sig_figs(0.0, 3)) # Expected: 0.0
# print(round_to_sig_figs(-17.6745, 3)) # Expected: -17.7

# Made with Gemini
````

# Section 1.4: Dimensional Analysis

Dimensional analysis is a way in mathematics for people to keep track of varying units in a problem. Here is an example of this:

$\frac{25m}{10m/s} = 2.5 s$

Here we have two different units (meters and meters/seconds) being used in this operation. Let's take a closer look at this:

$\frac{meters}{meters/seconds} = seconds$

The process of going from meters divided by meters per second can be expanded like this:

$\frac{meters}{meters/seconds} = \frac{meters}{meters * 1/seconds} = \frac{1}{1/seconds} = 1 * \frac{seconds}{1}$ = seconds
